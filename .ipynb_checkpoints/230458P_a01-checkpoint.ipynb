```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## Question 1: Piecewise Linear Transformations

```python
def intensity_transform(im, breakpoints):
    """
    Applies a piecewise linear intensity transformation to an image.
    """
    # Convert breakpoints to a numpy array for easy slicing
    breakpoints = np.array(breakpoints)
    x_coords = breakpoints[:, 0]
    y_coords = breakpoints[:, 1]
    
    # Create an array for all possible 256 input pixel values (0 to 255)
    x_vals = np.arange(256)
    
    # Use linear interpolation to calculate the corresponding output for every possible input
    # This creates a Lookup Table (LUT)
    lut = np.interp(x_vals, x_coords, y_coords).astype(np.uint8)
    
    # Apply the lookup table to the image efficiently using OpenCV
    transformed_im = cv2.LUT(im, lut)
    
    return transformed_im

# 1. Define the breakpoints based on Figure 1a (using 1-pixel offsets for vertical jumps)
breakpoints= [
    [0, 0],
    [70, 40],    # Pulls down the shadows (makes hair and dress richer)
    [128, 128],  # Anchors the exact middle gray so the image doesn't get too dark/light
    [190, 220],  # Pushes up the highlights (brightens the skin and eye reflections)
    [255, 255]   # Keeps pure white as pure white
]

# 2. Load the original image (Figure 1b) in grayscale
img = cv2.imread('Emma.png', cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: Could not load the image. Please check the filename and ensure it is in the same folder as this notebook.")
else:
    # 3. Apply the intensity transformation
    transformed_img = intensity_transform(img, breakpoints)
    
    # 4. Display the results side-by-side
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    
    # Plot the intensity transformation curve
    bp_array = np.array(breakpoints)
    axs[0].plot(bp_array[:, 0], bp_array[:, 1], color='maroon', linewidth=2)
    axs[0].set_title('Intensity Transformation')
    axs[0].set_xlabel('Input intensity')
    axs[0].set_ylabel('Output intensity')
    axs[0].set_xlim([0, 255])
    axs[0].set_ylim([0, 255])
    axs[0].grid(True, linestyle='--', alpha=0.6)
    
    # Plot the original image
    axs[1].imshow(img, cmap='gray', vmin=0, vmax=255)
    axs[1].set_title('Original Image')
    axs[1].axis('off')
    
    # Plot the transformed image
    axs[2].imshow(transformed_img, cmap='gray', vmin=0, vmax=255)
    axs[2].set_title('Transformed Image')
    axs[2].axis('off')
    
    plt.tight_layout()
    plt.show()
```

## Question 2: Targeted Intensity Mapping

```python
# 1. Define Breakpoints for Windowing
bp_white_matter = [
    [0, 0],
    [140, 0],
    [150, 255],
    [175, 255],
    [185, 0],
    [255, 0]
]

bp_gray_matter = [
    [0, 0],
    [95, 0],
    [105, 0],
    [175, 0],
    [185, 255],
    [210, 255],
    [220, 0],
    [255, 0]
]

# 2. Load the Brain Image
brain_img = cv2.imread('brain.png', cv2.IMREAD_GRAYSCALE)

if brain_img is None:
    print("Error: Could not load the brain image.")
else:
    # 3. Apply the transformations
    wm_img = intensity_transform(brain_img, bp_white_matter)
    gm_img = intensity_transform(brain_img, bp_gray_matter)
    
    # 4. Plotting
    fig, axs = plt.subplots(2, 3, figsize=(15, 10))
    x_vals = np.arange(256)
    
    # --- ROW 1: White Matter ---
    wm_arr = np.array(bp_white_matter)
    wm_lut = np.interp(x_vals, wm_arr[:,0], wm_arr[:,1])
    
    axs[0, 0].plot(x_vals, wm_lut, color='blue', linewidth=2)
    axs[0, 0].set_title('White Matter Transformation')
    axs[0, 0].set_xlabel('Input intensity')
    axs[0, 0].set_ylabel('Output intensity')
    axs[0, 0].set_xlim([0, 255])
    axs[0, 0].set_ylim([0, 255])
    axs[0, 0].grid(True, linestyle='--')
    
    axs[0, 1].imshow(brain_img, cmap='gray', vmin=0, vmax=255)
    axs[0, 1].set_title('Original Brain Image')
    axs[0, 1].axis('off')
    
    axs[0, 2].imshow(wm_img, cmap='gray', vmin=0, vmax=255)
    axs[0, 2].set_title('Accentuated White Matter')
    axs[0, 2].axis('off')
    
    # --- ROW 2: Gray Matter ---
    gm_arr = np.array(bp_gray_matter)
    gm_lut = np.interp(x_vals, gm_arr[:,0], gm_arr[:,1])
    
    axs[1, 0].plot(x_vals, gm_lut, color='red', linewidth=2)
    axs[1, 0].set_title('Gray Matter Transformation')
    axs[1, 0].set_xlabel('Input intensity')
    axs[1, 0].set_ylabel('Output intensity')
    axs[1, 0].set_xlim([0, 255])
    axs[1, 0].set_ylim([0, 255])
    axs[1, 0].grid(True, linestyle='--')
    
    axs[1, 1].imshow(brain_img, cmap='gray', vmin=0, vmax=255)
    axs[1, 1].set_title('Original Brain Image')
    axs[1, 1].axis('off')
    
    axs[1, 2].imshow(gm_img, cmap='gray', vmin=0, vmax=255)
    axs[1, 2].set_title('Accentuated Gray Matter')
    axs[1, 2].axis('off')
    
    plt.tight_layout()
    plt.show()
```

## Question 3: Gamma Correction in L*a*b* Space

```python
# 1. Load the image and convert from BGR (OpenCV default) to RGB (for Matplotlib)
img_bgr = cv2.imread('Legs.png')

if img_bgr is None:
    print("Error: Could not load the image.")
else:
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    
    # 2. Convert to L*a*b* color space
    lab_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2LAB)
    
    # 3. Split into L, a, and b planes
    l, a, b = cv2.split(lab_img)
    
    # 4. Apply Gamma Correction to the L plane
    gamma = 0.5  # <--- Here is the stated Gamma value
    
    # Create a lookup table (LUT) for efficient calculation
    x_vals = np.arange(256)
    lut_gamma = np.array([((i / 255.0) ** gamma) * 255 for i in x_vals]).astype(np.uint8)
    
    # Apply the LUT to the L plane
    l_corrected = cv2.LUT(l, lut_gamma)
    
    # 5. Merge the corrected L plane back with the original a and b planes
    lab_corrected = cv2.merge((l_corrected, a, b))
    
    # Convert back to RGB so we can display it properly
    img_corrected = cv2.cvtColor(lab_corrected, cv2.COLOR_LAB2RGB)
    
    # 6. Plotting: Display Images and Histograms as requested
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    
    # Original Image
    axs[0, 0].imshow(img_rgb)
    axs[0, 0].set_title('Original Image')
    axs[0, 0].axis('off')
    
    # Gamma Corrected Image
    axs[0, 1].imshow(img_corrected)
    axs[0, 1].set_title(f'Gamma Corrected Image (Gamma = {gamma})')
    axs[0, 1].axis('off')
    
    # Original L-Plane Histogram
    axs[1, 0].hist(l.ravel(), bins=256, range=[0, 256], color='gray')
    axs[1, 0].set_title('Histogram of Original L Plane')
    axs[1, 0].set_xlabel('Pixel Intensity (0-255)')
    axs[1, 0].set_ylabel('Frequency')
    axs[1, 0].grid(True, linestyle='--', alpha=0.6)
    
    # Corrected L-Plane Histogram
    axs[1, 1].hist(l_corrected.ravel(), bins=256, range=[0, 256], color='black')
    axs[1, 1].set_title(f'Histogram of Corrected L Plane (Gamma = {gamma})')
    axs[1, 1].set_xlabel('Pixel Intensity (0-255)')
    axs[1, 1].set_ylabel('Frequency')
    axs[1, 1].grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()
```

To enhance the image, a gamma value of $\gamma = 0.5$ was applied to the L plane. Because this value is less than 1, it successfully brightens the underexposed shadow regions (the dark rocks) while leaving the color properties (the a and b planes) completely intact. As seen in the histograms, the pixel frequencies have been stretched and shifted toward the higher (brighter) intensity ranges

## Question 4: Vibrance Enhancement in HSV Space

```python
# 1. Load the image
img_bgr = cv2.imread('spiderman.png')

if img_bgr is None:
    print("Error: Could not load the image.")
else:
    # Convert BGR to RGB for correct display in Matplotlib
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # 2. Convert to HSV and split into H, S, and V planes (Task a)
    hsv_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)
    h, s, v = cv2.split(hsv_img)

    # 3. Define the transformation parameters
    a = 0.5      # Task c: Adjust this value between 0 and 1!
    sigma = 70   # Task b: Given in the assignment

    # Create the Lookup Table (LUT) for the saturation transformation
    x_vals = np.arange(256)
    
    # The mathematical formula provided in the assignment
    transformation = x_vals + a * 128 * np.exp(-((x_vals - 128) ** 2) / (2 * sigma ** 2))
    
    # Use np.clip to act as the "min(..., 255)" part of the formula, keeping values within 0-255
    lut_saturation = np.clip(transformation, 0, 255).astype(np.uint8)

    # 4. Apply the intensity transformation to the saturation plane (Task b)
    s_enhanced = cv2.LUT(s, lut_saturation)

    # 5. Recombine the three planes (Task d)
    hsv_enhanced = cv2.merge((h, s_enhanced, v))
    
    # Convert back to RGB to view the final result
    img_enhanced = cv2.cvtColor(hsv_enhanced, cv2.COLOR_HSV2RGB)

    # 6. Display the original, enhanced image, and transformation curve (Task e)
    fig, axs = plt.subplots(1, 3, figsize=(18, 5))

    # Original Image
    axs[0].imshow(img_rgb)
    axs[0].set_title('Original Image')
    axs[0].axis('off')

    # Enhanced Image
    axs[1].imshow(img_enhanced)
    axs[1].set_title(f'Vibrance Enhanced (a = {a})')
    axs[1].axis('off')

    # Transformation Curve
    axs[2].plot(x_vals, lut_saturation, color='orange', linewidth=2)
    # Adding a faint reference line showing what "no change" would look like
    axs[2].plot(x_vals, x_vals, color='gray', linestyle='--', alpha=0.5) 
    axs[2].set_title('Saturation Intensity Transformation')
    axs[2].set_xlabel('Input Saturation')
    axs[2].set_ylabel('Output Saturation')
    axs[2].set_xlim([0, 255])
    axs[2].set_ylim([0, 255])
    axs[2].grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()
```

## Question 5: Global Histogram Equalization

```python
def custom_hist_equalization(img):
    """
    Performs histogram equalization on a grayscale image from scratch.
    """
    # 1. Calculate the histogram (pixel counts for 0-255)
    hist, bins = np.histogram(img.flatten(), 256, [0,256])
    
    # 2. Calculate the Cumulative Distribution Function (CDF)
    cdf = hist.cumsum()
    
    # 3. Normalize the CDF to be strictly between 0 and 255
    # We ignore 0 values temporarily to avoid dividing by zero issues
    cdf_masked = np.ma.masked_equal(cdf, 0)
    cdf_normalized = (cdf_masked - cdf_masked.min()) * 255 / (cdf_masked.max() - cdf_masked.min())
    
    # Fill the masked zero values back in and convert to 8-bit integers
    cdf_final = np.ma.filled(cdf_normalized, 0).astype('uint8')
    
    # 4. Map the original image pixels to their new equalized values using the CDF
    equalized_img = cdf_final[img]
    
    return equalized_img

# 1. Load the original image (Figure 5) in grayscale
img_pollen = cv2.imread('pebbles.png', cv2.IMREAD_GRAYSCALE)

if img_pollen is None:
    print("Error: Could not load the image. Check the filename.")
else:
    # 2. Apply our custom histogram equalization function
    img_equalized = custom_hist_equalization(img_pollen)
    
    # 3. Display the results and the histograms
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))
    
    # Original Image
    axs[0, 0].imshow(img_pollen, cmap='gray', vmin=0, vmax=255)
    axs[0, 0].set_title('Original Image (Fig 5)')
    axs[0, 0].axis('off')
    
    # Equalized Image
    axs[0, 1].imshow(img_equalized, cmap='gray', vmin=0, vmax=255)
    axs[0, 1].set_title('Histogram Equalized Image')
    axs[0, 1].axis('off')
    
    # Original Histogram
    axs[1, 0].hist(img_pollen.ravel(), bins=256, range=[0, 256], color='gray')
    axs[1, 0].set_title('Original Histogram')
    axs[1, 0].set_xlabel('Pixel Intensity')
    axs[1, 0].set_ylabel('Frequency')
    axs[1, 0].grid(True, linestyle='--', alpha=0.6)
    
    # Equalized Histogram
    axs[1, 1].hist(img_equalized.ravel(), bins=256, range=[0, 256], color='black')
    axs[1, 1].set_title('Equalized Histogram')
    axs[1, 1].set_xlabel('Pixel Intensity')
    axs[1, 1].set_ylabel('Frequency')
    axs[1, 1].grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()
```
